#### Plotting out the factors and RMS

```python
import numpy as np
import matplotlib.pyplot as plt
factors=np.array([1.5,1.72,1.94,2.16,2.38,2.61,2.83,3.05,3.27,3.5])
RMS=np.array([[0.0189,0.0169,0.0149,0.0139,0.0141,0.0132,0.0136,0.0144,0.0172,0.0186],[0.0202,0.0187,0.0164,0.0147,0.0147,0.0143,0.0141,0.0154,0.0199,0.0307] ])
fig, ax = plt.subplots(1,1,figsize=(14, 6))
ax.plot(factors,RMS[0],'rs',label="Alt")
ax.plot(factors,RMS[1],'bs',label='Azm')
ax.set_xlabel("Corrective factor")
ax.set_ylabel("RMS in degrees")
ax.legend()
fig.suptitle("RMS error vs corrective factor")
plt.show()
```
